```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python

from sqlalchemy import create_engine, URL
import getpass

username = "postgres"
password = getpass.getpass("Enter PostgreSQL password: ")

host = "localhost"
port = 5432
database = "customer_behaviour"

# Create a safe PostgreSQL connection
url = URL.create(
    "postgresql+psycopg2",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(url)

# Test connection
with engine.connect() as connection:
    print("Connected to PostgreSQL successfully!")

# Upload data
df.to_sql(
    "customer",
    engine,
    if_exists="replace",
    index=False
)

print("Data uploaded successfully!")
```
